# Workbook Quy trình Xử lý Dữ liệu
## Bước 1: Thu thập và lọc dữ liệu ngành hàng Máy lạnh treo tường

```python
import pandas as pd
df = pd.read_json('data/raw/tiki_products_60k.jsonl', lines=True)
df.shape
```

```python
# Xem thử 5 dòng đầu và các cột của bộ dữ liệu
df.head()
```

```python
# Thống kê số lượng sản phẩm theo ngành hàng để chọn ngành hàng phân tích
so_luong_nganh = df['primary_category_name'].value_counts()
so_luong_nganh[so_luong_nganh >= 150].head(15)
```

```python
# Lọc ra ngành hàng 'Máy lạnh treo tường'
df_may_lanh = df[df['primary_category_name'] == 'Máy lạnh treo tường'].copy()
df_may_lanh = df_may_lanh.reset_index(drop=True)

print('Số dòng, số cột:', df_may_lanh.shape)
print('Số thương hiệu:', df_may_lanh['brand_name'].nunique())
df_may_lanh.head()
```

```python

df_may_lanh.to_csv('data/raw/tiki_may_lanh_229.csv', index=False, encoding='utf-8-sig')
```
